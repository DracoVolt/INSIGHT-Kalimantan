# 03 CA prototype

This notebook explores the cellular automaton used for the demo.

```python
from src.config import bounding_box, grid_spacing_deg, default_wind_direction_deg
from src.ca import CaGrid, CaParams, run_simulation, seed_cell, step_explanation
from src.fire_data import build_cell_grid as build_grid

bbox = bounding_box()
grid = CaGrid().initialize(
    build_grid(bounding_box=bbox).n_rows,
    build_grid(bounding_box=bbox).n_cols,
)

wind = default_wind_direction_deg()
params = CaParams(wind_dir_deg=wind)

center_row = grid.n_rows // 2
center_col = grid.n_cols // 2
seed_cell(grid, center_row, center_col)

frames = run_simulation(grid, steps=10, params=params)
for i, frame in enumerate(frames):
    summary = frame.summary()
    print(f"Step {i}: burning={summary.get('burning')}, burned={summary.get('burned')}, healthy={summary.get('healthy')}")

print()
print(step_explanation(params))
```
